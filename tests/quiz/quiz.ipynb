GitHub
1.1. Что такое GitHub? GitHub - это самый популярный хостинг-социальная сеть 
опен сурс git-репозиториев в мире, позволяющий упрощать и организовывать 
работу с кодом разработчикам, находящимся как в команде, так и одиночкам. 

1.2. Как GitHub связан с Git? Github является онлайн платформой, в том время
как git расположен локально на "машине" и сам по себе является инструментом 
версионирования, с помощью которого также можно взаимодействовать с гитхаб.

1.3. Чем отличается fork репозитория от его клонирования (clone)? fork
"копирует" чужой репозиторий в ваш собственный, создавая собственную копию
проекта в вашем аккаунте, c которой можно делать что угодно. В то же время, 
клонирование лишь локально загружает файлы из указанного репозитория 
(не обязательно своего) на ваш компьютер.

1.4. Зачем нужны и как работают pull requests? PR-s нужны для того, чтобы
согласованно делать изменения в репозитории, связанные с главной основной 
версией проекта (обычно ветка main). Описание процесса: программист выполняет 
определенную задачу, связанную с репозиторием (фикс бага, дополнение
документации, новая фича и тд) на отдельной ветке, делает коммиты. В момент, 
когда выполняемая задача принимает законченный вид, программист создает PR, 
чтобы объединить изменения (коммиты) из своей ветки с главной версией проекта 
(main). Когда пул-реквест успешно создан, ответственные за процесс его 
проверки решают закрыть пул-реквест (отвергнуть) или выполнить merge - слияние 
веток, иногда делая для этого code review и тесты. При слиянии веток сделанные 
изменения попадают в основную версию программы.

1.5. GitHub использует ваш почтовый адрес для привязки ваших Git коммитов к

 почта и имя автора (локально) и, при взаимодествии с гитхаб, гит ищет гитхаб
аккаунт, привязанный к этой почте

2.9. Как открыть запрос слияния, указывающий на другой запрос слияния, и зачем 
это нужно?

**Ответ:** Можно создать PR, указав в качестве целевой (`base`) ветки ветку 
другого PR, а не `main`. Это полезно для разделения больших изменений на 
несколько зависимых PR (stacked PR), их независимого ревью, а также для 
предложения исправлений в чужой PR.

**Механизм:** Допустим, PR №1 предлагает слить `feature-A` в `main`. Создаём 
новую ветку `feature-B` на основе `feature-A`, вносим изменения и открываем PR 
№2, у которого `base = feature-A`, а `head = feature-B`. После слияния PR №2 
изменения попадут в `feature-A` и автоматически отразятся в PR №1.

Если ветка первого PR находится в чужом форке, можно создать PR с целевой 
веткой в этом форке, используя собственный форк для изменений. Таким образом, 
PR технически направляется не на другой PR, а на его исходную ветку.

Создание запроса на слияние
4 Можно ли открыть пул реквест, если вы ничего не вносили в FORK? нет
4.1 Что нужно сделать чтобы открыть пул реквест? (Рисунок 90) После пуша 
коммитов в тематическую ветку гитхаб предложит создать пул реквест прямо на 
главной странице репозитория, также можно создать его из ветки, из которой 
планировались изменения. Для создания пул реквеста необходимо сделать описание 
(желательно подробно), дать название PR и нажать зеленую кнопку create pull 
request
4.2 Что нужно сделать Если ваш Форк устарел? Fetch upstream - загрузить 
изменения оригинального репозитория (только скачать), а потом влить их в свою 
ветку.
Через командную строку:
1) git remote add upstream <ссылка_на_оригинальный_репозиторий>
2) git fetch upstream
3) git checkout main
4) git merge upstream/main
5) git push origin main

4.3 Что нужно сделать если в пул реквесте имеются конфликты слияния 
(Рисунок 96) Следует изменить свою ветку так, чтобы исключить конфликты. 
Существует два основных варианта это сделать: 
1) Перебазировать свою ветку относительно целевой (git rebase) — этот способ 
делает историю линейной и чистой, но он сложнее в исполнении и может стать 
источником ошибок.
2) Слить целевую ветку в свою (git merge) — этот вариант используется чаще 
всего. Он намного проще, безопаснее и позволяет разрешить все конфликты за 
один раз, хотя и делает историю чуть более подробной из-за появления 
мерж-коммита.Процесс для популярного (второго) варианта при работе с форком:
Нужно подключить исходный (оригинальный) репозиторий как новый удаленный 
источник (upstream), скачать свежие данные и слить изменения из его основной 
ветки в свою тематическую ветку. Если возникнут конфликты — вручную исправить 
проблемные строчки кода в файлах, сделать коммит слияния и отправить (git push)
 изменения в свой форк на GitHub. После этого конфликт в Pull Request исчезнет,
  и его можно будет успешно завершить.

Особенные файлы
9 Что такое и зачем нужен файл README? файл ридми - это файл с названием и 
расширением README.md, он предназначен для отображения основной информации об 
репозитории для пользователя в презентабельном виде (именно содержимое 
отображается на странице репозитория). Для написания данного файла используется
язык markdown. Обычно в нем указываются контакты авторов, описание основного 
функционала программы, инструкции по установке запуску и использованию, стек 
технологий, примеры использования и тд
9.1 Что такое и зачем нужен файл CONTRIBUTING (Рисунок 122) этот файл будет 
отображаться при попытке сделать PR. Он нужен для того чтобы информировать 
потенциальных контрибьютеров о требованиях и правилах, связанных с 
пул-реквестом. Возможное содержимое включает в себя: требования к коммитам, 
оформлению кода; инструкция по тестированию, шаблон для создания ишьюса или PR. 

Резюмируя можно сказать, что readme отвечает на вопросы типа как пользоваться 
данным ПО, для чего оно нужно, то есть предназначен больше для пользователя
CONTRIBUTING отвечает на вопросы по типу: что нужно сделать, чтобы авторы 
репозитория приняли в рассмотрение мой вклад в проект, то есть он предназначен 
для контрибьюторов
