Общие вопросы
    1. Что такое Issues на GitHub и для чего они используются? это гибкий 
    способ коммуникации на гитхаб, который позволяет сообщать о проблемах 
    связанных с ПО в репозитории (в широком смысле, не только с кодом), 
    предлагать новые идеи для улучшения продукта и просто обсуждать его.
    2. Чем Issues отличаются от других инструментов управления задачами? их 
    главные отличия - удобная привязка к коду и Git, которая позволяет быстро 
    перейти к проблемному коду в репозитори и автоматически закрыть issues при 
    исправлении проблемы, графически отобразив исправления в истории 
    репозитория, также можно отметить удобную организацию (теги, майлстоуны, 
    всевозможные фильтры для поиска ишьюсов и тд)
    3. Какие основные компоненты (поля) есть у каждого Issue? title, 
    description, Assignees, Labels, Projects, Milestone, Development 
    (связанные PR), а также комментарии и статус (open/closed).

Создание Issues
    4. Как создать новое Issue в репозитории? зайти в раздел issues в верхнем 
    баре и нажать  кнопку new issue (также есть способы: из CLI, из шаблона 
    проекта, из строки кода -> reference in new issue)
    5. Какие данные рекомендуется указывать в описании Issue для лучшего 
    понимания задачи? трассировку из терминала; проблемный файл (путь к нему, 
    его название), код ошибки
    6. Какие теги (labels) можно добавить к Issue? Какие из них стандартные? 
    bug, documentation, duplicate, enhancement, good first issue, help wanted, 
    invalid, question, wontfix, также можно создать свои кастомные тэги. 
    Стандартные - все 9 перечисленных первыми.
    7. Как прикрепить Assignees (ответственных) к Issue? в правой части экрана,
     в самом верху столбца - assignees -> нажать шестеренку и выбрать 
     ответственных из выпадающего списка (также можно быстро отметить себя -  
     «assign yourself»), ограничение - до 10 человек

Работа с Issues
    8. Как использовать Labels для классификации задач? Выбрать самый 
    подходящий label под конкретную задачу, например, при исправлении бага, в 
    ишьюс добавляется лейбл bug. Также можно создать свой (кастомный) под 
    определенные типы задач. Вот несколько примеров распространенных меток в 
    современной разработке: area: frontend, area: backend, area: API. Нужно 
    отметить, что современные CI/CD системы могут использовать лейблы для 
    автоматизации сценариев (например, рассылка ботом ишьюсов и лейблом prio: 
    high). Такая удобная классификация позволяет разработчикам очень быстро 
    находить необходимые ишьюсы.
    9. Для чего нужен Milestone, и как связать его с Issue? Майлстоун 
    позволяет привязать ишьюс к временному отрезку, отслеживать шкалу 
    прогресса в процентах (сколько ишьюсов закрыто внутри вехи), помогает в 
    случае надобности отыскать соответствующий ишьюс. Пример использования: 
    объединить множество мелких issues (bug, documentation, feature) под одной 
    целью: "релиз версии 2.0", при этом обозначив дедлайн.  Чтобы связать его с
     issue, нужно зайти в сам ишьюс и в правом боковом меню нажать на 
     шестеренку, где из выпадающего списка выбрать нужную веху.
    10. Как привязать Issue к пул-реквесту (Pull Request)? в description к PR 
    надо написать ключевое слово и ссылку на issues, например closes 
    ССЫЛКА_НА_ИШЬЮС (или просто номер проблемы с решеткой: #number_of_issue). 
    Главный плюс - при выполнении слияния веток (мерж) ишьюс закроется 
    автоматически. Альтернативный вариант: в пул-реквесте в боковом меню 
    нажимаем шестеренку рядом с development и выбираем нужный ишьюс  
    11. Как добавить комментарий к существующему Issue? Листать вниз до того 
    момента, как встретится поле для комментария и надпись  над ней add a 
    comment, написать в markdown комментарий, а затем отправить, нажав на 
    зеленую кнопку comment (либо нажав ctrl + enter)

Закрытие и завершение Issues
    12. Как закрыть Issue вручную? Под комментариями кнопка close issue
    13. Можно ли автоматически закрыть Issue с помощью сообщения в коммите 
    или пул-реквесте? Как это сделать? да (даже несколько сразу), использовать 
    ключевые слова на подобие: close, closes, closed, fix, fixes, fixed, 
    resolve, resolves, resolved и номер проблемы с решеткой или ссылку, при 
    этом она закроется не сразу после коммита, а только после слияния (merge) 
    с главной веткой main
    14. Как повторно открыть закрытое Issue, если работа ещё не завершена? в 
    том же месте, где он был закрыт - открыть: reopen issue

Фильтрация и поиск
    15. Как найти все открытые или закрытые Issues в репозитории? во вкладке 
    issues на панели под all issues -> open или closed при нажатии покажут 
    соответствующие ишьюсы. Можно также использовать строку поиска над меню, 
    где команды на подобие is:issue is:open
    16. Как использовать фильтры для поиска Issues по меткам, исполнителям или 
    другим критериям? на этой же панели (из прошлого вопроса) нажимаем на 
    соответствующую надпись и выбираем необходимое в выпадающем меню 
    (или использовать поисковые запросы)
    17. Как сортировать Issues по приоритету, дате создания или другим 
    параметрам? опять эта же панель - самая левая из надписей -> выбираем 
    соответствующую опцию из выпадающего меню (или используем поиск). В 
    стандартных issue на гитхаб нет поля "приоритет", для того, чтобы 
    организовать такую сортировку, нужно либо использовать labels (prio:high, 
    prio:low), либо через гитхаб проекты

Интеграции и автоматизация
    18. Как настроить автоматические уведомления о новых или изменённых Issues?
     в правом столбце -> под notifications: subscribe/unsubscribe. В 
     репозитории: watch (со значком глазика) -> custom -> чекбокс с issues, 
    19. Что такое Projects в контексте GitHub, и как связать их с Issues? 
    projects - это инструмент, позволяющий более эффективно работать в команде,
     отслеживать задачи, время, баги, разбивать работу на спринты (итерации), 
     визуализировать процессы с помощью канбан доски и тд. Связать с ишьюсом 
     можно в правой панели -> шестеренка справа от надписи projects. Также 
     можно добавить из проекта непосредственно -> add item #num_of_issue
    20. Какие сторонние инструменты можно использовать для автоматизации 
    работы с Issues (например, боты, Webhooks)? GitHub Webhooks и GitHub 
    Apps,Make (бывший Integromat) / Zapier / n8n (позволяют настраивать 
    цепочки действий с базами данных, таблицами, чатами и тд), GitHub Actions 
    (Встроенные CI/CD роботы), Jira / Asana / ClickUp (глубокая интеграция, 
    передают информацию о выполненных задачах сразу менеджеру)

Коллаборация
    21. Как упомянуть другого пользователя в комментарии к Issue? использовать 
    конструкцию вида: @nickname_on_GH 
    22. Как запросить дополнительные данные или уточнения у автора Issue? 
    Специального алгоритма вроде нет, но самый логичный вариант: написать 
    комментарий с необходимым запросом и тэгнуть автора, также можно сделать 
    issue с кастомными тегами по типу needs-info
    23. Что делать, если Issue неактуально или его нужно объединить с другим? 
    Если нужно закрыть issue, то лучше оставить развернутый комментарий с 
    причинами закрытия, выбрать причину (close as not planned), повесить тег 
    (например, wontfix, invalid). Если необходимо объединить с другим 
    (дубликаты), тогда нужно оставить комментарий с ссылкой на оригинал 
    (Пример комментария: Duplicate of #42). Примечание: у дубликатов в GitHub 
    есть отдельная причина закрытия «Close as duplicate».

Практические аспекты
    24. Как использовать шаблоны для создания Issues? Настраиваем шаблоны: 
    настройки репозитория -> features -> issues -> set up templates -> выбрать 
    пресеты и отредактировать их под себя. Пользователь, создавая ишьюс может 
    выбрать нужный ему за ранее настроенный разработчиками репозитория шаблон, 
    где будут прописана вся необходимая информация и поля с вопросами, которые 
    нужно заполнить. Помимо предзаполненного текста с запрашиваемыми полями 
    (который избавляет от бесполезных, малоинформативных сообщений) можно также
     настроить автоматическое назначение ответственного и лейблы.
    25. Что такое Linked Issues, и как создать связь между задачами? Это 
    механизм гитхаба, который позволяет связывать ишьюсы друг с другом и ишьюсы
     с пул-реквестами. Чтобы создать связь одного ишьюса с другим нужно в 
     комментарии либо выбрать через reference, либо просто написать номер 
     ишьюса с решеткой
    26. Какие метрики (например, время выполнения) можно отслеживать с помощью 
    Issues? Lead time (полное время выполнения с момента открытия ишьюса до его
     закрытия), cycle time (время непосредственно в работе над исправлением), 
     time to first response (время первого до первого ответа, понятно по 
     названию), velocity (количество закрытых задач за определенный срок), 
     defect density (плотность дефектов - соотношение кол-ва баговых ишьюсов к 
     кол-ву ишьюсов с новыми фичами - помогает понять на что команда тратит 
     больше времени, над исправлениями или новыми фичами), activity/engagement 
     (уровень активности в обсуждении). Посмотреть их можно на вкладке insights
      и projects 
    27. Какие best practices рекомендуются при работе с Issues в команде?
    1) Всегда использовать шаблоны для issues, разделив их как минимум на 3 
    категории: Bug report (ошибки), Feature request (идеи) и Task (рутинные 
    задачи)
    2) один ишьюс - одна проблема
    3) понятные, информативные заголовки 
    4) назначение ответственных 
    5) использование майлстоунов и канбан-досок
    6) использовать понятную систему лейблов с интуитивными цветами (красный - 
    баг, высокий приоритет), (зеленый - good first issue)
    7) фиксировать изменения для команды в комментариях ишьюса (логика 
    выполнения задачи)
    8) правильно закрывать задачи (тип закрытия, (не) запланированное закрытие 
    + комментарий с пояснением причин)
    9) автоматизированное закрытие через коммиты (fixes #16)
    10) чистка старых задач, например настроить бота через github actions, 
    который будет автоматически тегать или закрывать задачи, которым больше 
    двух-трех месяцев
